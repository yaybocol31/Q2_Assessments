```python

# Activity 3: Flower Shop Data Analyzer

flowers = ["Rose", "Tulip", "Daisy", "Lily", "Sunflower"]

def display():
    print("\nFlower Inventory:")
    for i, flower in enumerate(flowers):
        print(i, "-", flower)

def search():
    name = input("Enter flower to search: ")
    if name in flowers:
        print(name, "was found!")
    else:
        print(name, "was not found.")

def compute():
    print("Total flower types:", len(flowers))

def update():
    index = int(input("Enter index to update: "))
    if 0 <= index < len(flowers):
        name = input("Enter new flower: ")
        flowers[index] = name
        print("Flower updated!")
    else:
        print("Invalid index!")

def menu():
    while True:
        print("\n=== Flower Shop Data Analyzer ===")
        print("1. Display Flowers")
        print("2. Search Flower")
        print("3. Count Flower Types")
        print("4. Update Flower")
        print("5. Display Updated Flowers")
        print("6. Exit")

        choice = input("Enter your choice: ")

        if choice == "1":
            display()
        elif choice == "2":
            search()
        elif choice == "3":
            compute()
        elif choice == "4":
            update()
        elif choice == "5":
            display()
        elif choice == "6":
            print("Thank you!")
            break
        else:
            print("Invalid choice!")

menu()
```
