```python

# FA2: Process Your Data (hi)

games = [
    "The Man From The Window",
    "Deadplate",
    "The Coffin of Andy And Leyley",
    "Little Goody Two Shoes",
    "Little Miss Fortune"
]

print("Original Game List:", games)

index = int(input("Input Index to Update: "))
new_game = input("Input New Game: ")

start = int(input("Slice Start: "))
end = int(input("Slice End: "))

sliced = games[start:end]
print("Sliced Game List:", sliced)

search = input("Enter game to search: ")

if search in sliced:
    print(search, "was found!")
else:
    print(search, "was not found.")

games[index] = new_game

print("Updated Game List:", games)
```
